```python
import math
import random


def calculate_conflicts(board):
    """Counts the number of pairs of queens attacking each other."""
    conflicts = 0
    n = len(board)
    for i in range(n):
        for j in range(i + 1, n):
            # Same row or same diagonal
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1
    return conflicts


def get_neighbor(board):
    """Generates a neighbor state by moving one random queen to a new row."""
    neighbor = list(board)
    n = len(board)
    col = random.randint(0, n - 1)
    new_row = random.randint(0, n - 1)
    while new_row == neighbor[col]:
        new_row = random.randint(0, n - 1)
    neighbor[col] = new_row
    return neighbor


def simulated_annealing(
    n=8, initial_temp=100.0, cooling_rate=0.99, min_temp=0.001, max_steps=100000
):
    # Start with one queen per column in a random row
    current_board = [random.randint(0, n - 1) for _ in range(n)]
    current_conflicts = calculate_conflicts(current_board)
    temp = initial_temp

    for step in range(max_steps):
        if current_conflicts == 0:
            return current_board, step

        neighbor = get_neighbor(current_board)
        neighbor_conflicts = calculate_conflicts(neighbor)
        delta = neighbor_conflicts - current_conflicts

        # Accept better moves, or worse moves with probability exp(-delta / T)
        if delta < 0 or random.random() < math.exp(-delta / temp):
            current_board = neighbor
            current_conflicts = neighbor_conflicts

        temp = max(temp * cooling_rate, min_temp)

    return None, max_steps


def print_board(board):
    n = len(board)
    for row in range(n):
        line = " ".join("Q" if board[col] == row else "." for col in range(n))
        print(line)


if __name__ == "__main__":
    solution, steps = simulated_annealing()
    if solution:
        print(f"Solution found in {steps} steps: {solution}\n")
        print_board(solution)
    else:
        print("Failed to find a solution within the step limit.")
```
