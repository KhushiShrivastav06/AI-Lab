```python
def vacuum_cleaner():

    # Initial state of rooms
    room = {
        "A": "Dirty",
        "B": "Dirty"
    }

    # Vacuum starts from Room A
    position = "A"

    # Cost of actions
    suck_cost = 1
    move_cost = 1

    total_cost = 0

    print("----- Vacuum Cleaner Problem -----")

    while True:

        print("\nVacuum is in Room:", position)
        print("Room A:", room["A"])
        print("Room B:", room["B"])

        # If current room is dirty
        if room[position] == "Dirty":
            print("Action: SUCK")
            room[position] = "Clean"

            total_cost += suck_cost
            print("Cost:", suck_cost)

        # If current room is clean and vacuum is in A
        elif position == "A":
            print("Action: MOVE RIGHT")
            position = "B"

            total_cost += move_cost
            print("Cost:", move_cost)

        # If current room is clean and vacuum is in B
        elif position == "B":
            print("Action: MOVE LEFT")
            position = "A"

            total_cost += move_cost
            print("Cost:", move_cost)

        # Check whether both rooms are clean
        if room["A"] == "Clean" and room["B"] == "Clean":
            print("\nBoth rooms are CLEAN!")
            print("Vacuum Cleaner stopped.")
            print("Total Cost:", total_cost)
            break


# Call the function
vacuum_cleaner()
```
