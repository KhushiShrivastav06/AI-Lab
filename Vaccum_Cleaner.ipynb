```python
def vacuum_cleaner():
    room = {
        "A": "Dirty",
        "B": "Dirty"
    }
    position = "A"
    suck_cost = 1
    move_cost = 1

    total_cost = 0

    print("----- Vacuum Cleaner Problem -----")

    while True:

        print("\nVacuum is in Room:", position)
        print("Room A:", room["A"])
        print("Room B:", room["B"])

        if room[position] == "Dirty":
            print("Action: SUCK")
            room[position] = "Clean"

            total_cost += suck_cost
            print("Cost:", suck_cost)

        elif position == "A":
            print("Action: MOVE RIGHT")
            position = "B"

            total_cost += move_cost
            print("Cost:", move_cost)

        elif position == "B":
            print("Action: MOVE LEFT")
            position = "A"

            total_cost += move_cost
            print("Cost:", move_cost)

        if room["A"] == "Clean" and room["B"] == "Clean":
            print("\nBoth rooms are CLEAN!")
            print("Vacuum Cleaner stopped.")
            print("Total Cost:", total_cost)
            break

vacuum_cleaner()
```
